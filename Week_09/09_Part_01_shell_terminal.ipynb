## 1. Terminal and shell

The **terminal** is the window. The **shell** is the program inside it that reads your commands. It is also a language, like Python and R. The shell has been around longer than most of its users have been alive, and it survives because a few keystrokes can automate repetitive work and chain small tasks into bigger workflows.

| | Mac | Windows |
|---|---|---|
| App to open | **Terminal** (`Cmd`+`Space`, type *Terminal*) | **PowerShell** (Start menu, type *PowerShell*) |
| Shell | zsh | PowerShell |
| The prompt looks like | `student@MacBook ~ %` | `PS C:\Users\student>` |
| Root of all folders | `/` | `C:\` |
| Your home folder, `~` | `/Users/student` | `C:\Users\student` |

Don't type the prompt itself (`%` or `PS ...>`). Type only the command after it.

## 3. Commands

Same name in both shells? Great. Where they differ, use your column.

### Core commands (quiz)

| Mac (zsh) | Windows (PowerShell) | What it does |
|---|---|---|
| `whoami` | `whoami` | Who am I? Prints your username |
| `echo "hi"` | `echo "hi"` | Print text |
| `pwd` | `pwd` | Print working directory: where am I? |
| `ls`<br>`ls -a` | `ls`<br>`ls -Force` | List files<br>…including hidden ones |
| `cd data` | `cd data` | Change directory |
| `mkdir data`<br>`mkdir data docs` | `mkdir data`<br>`mkdir data, docs` | Make a folder<br>…or several at once (PowerShell needs commas) |
| `rm notes.txt`<br>`rm -r data` | `rm notes.txt`<br>`rm data -Recurse` | Remove a file, **forever** (no Trash)<br>…or a folder and everything in it |
| `rmdir data` | `rmdir data` | Remove an **empty** folder. ⚠️ PowerShell asks, then deletes a non-empty one too |
| `mv old.txt new.txt` | `mv old.txt new.txt` | Move or rename. zsh overwrites silently (add `-i` to be asked); PowerShell refuses unless you add `-Force` |
| `cp a.txt b.txt`<br>`cp -r data backup` | `cp a.txt b.txt`<br>`cp data backup -Recurse` | Copy; overwrites an existing file<br>…or a whole folder |

### Working with files

| Mac (zsh) | Windows (PowerShell) | What it does |
|---|---|---|
| `touch notes.txt` | `New-Item notes.txt` | Create a blank file |
| `nano notes.txt` | `notepad notes.txt` | Edit a file in a simple text editor |
| `cat notes.txt` | `cat notes.txt` | Print a file |
| `less big.txt` | `more big.txt` | Scroll through a file; `q` to quit |
| `head -n 3 data.csv` | `Get-Content data.csv -Head 3` | First 3 lines (`head` alone shows 10) |
| `tail -n 3 data.csv` | `Get-Content data.csv -Tail 3` | Last 3 lines |
| `wc -l data.csv` | `(Get-Content data.csv).Count` | Count lines |
| `sort -n nums.txt` | <code>Get-Content nums.txt &#124; Sort-Object {[int]$_}</code> | Sort lines as numbers |
| `uniq names.txt` | <code>Get-Content names.txt &#124; Get-Unique</code> | Drop repeated lines that sit next to each other |
| `cut -d, -f2 data.csv` | <code>Import-Csv data.csv &#124; Select-Object school</code> | Pull out one column (by number in zsh, by name in PowerShell) |
| `diff a.txt b.txt` | `Compare-Object (Get-Content a.txt) (Get-Content b.txt)` | Show differences between two files |
| `open .`<br>`open notes.txt` | `start .`<br>`start notes.txt` | Open the folder in Finder / File Explorer<br>…or a file in its default app |
| `history` | `history` | Numbered list of recent commands |
| `!76` | `r 76` | Run command number 76 from `history` again |

## 4. Where PowerShell is different

- **Unix-style options often fail.** `ls -la`, `rm -rf`, and `cp -i` don't work. PowerShell spells options out: `-Force`, `-Recurse`.
- **Lists use commas**, not spaces: `mkdir data, docs, output`.
- **Some Mac commands don't exist**, such as `touch`, `nano`, `head`, `wc`, and `open`. Use the PowerShell column above.

## 5. File names: no spaces!

Try this once:

1. In Finder or File Explorer, make a folder named `data science` in your home folder.
2. Try to remove it from the command line:

```
cd ~
ls
rmdir data science
```

It fails, because the shell reads `data` and `science` as two separate names. Quote the name, or type `rmdir data` and press `Tab` to let the shell fill it in:

```
rmdir "data science"
```

**Name files so you never need quotes:**

- No spaces. Use snake_case (`shell_warmup_activities`) or camelCase (`shellWarmupActivities`)
- Prefer lower case, especially the first letter. It's faster with `Tab`
- Use plain letters, digits, `_`, and `-` only. Never `/` or `\`, and don't start a name with `-`
- Avoid extra dots. A dot marks the file extension, as in `.csv` or `.ipynb`
